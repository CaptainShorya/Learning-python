### Reading Data From Different Sources


```python
import pandas as pd
from io import StringIO
# Many pandas functions are written to read from: files (.csv, .json),file paths,file-like objects
#StringIO converts a string into an in-memory file-like object so that pandas functions can read it as if it were a file.
Data = '{"employee_name": "James", "email": "james@gmail.com", "job_profile": [{"title1":"Team Lead", "title2":"Sr. Developer"}]}' ##String
print(type(Data)) 
df=pd.read_json(StringIO(Data)) #read_json() -> Reads the JSON data and converts it into a DataFrame.pandas interprets it as columns with values.
# read_json() expects a file or file-like input
```

```python
df
# orient decides how rows and columns are represented in JSON.
```

```python
df.to_json() # Converts a pandas DataFrame into JSON format(Json String). 
# Default: orient="columns" -> So pandas organizes data column-wise.
```

```python
df.to_json(orient='index') # Index becomes the key
```

```python
df.to_json(orient='records') #row-wise (most common)  #It returns one JSON string,not a python list
```

```python
##Reading data from specefic url(if in form of CSV) 
df=pd.read_csv("https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data",header=None) ## df -> dataframe
```

```python
url="https://en.wikipedia.org/wiki/Mobile_country_code"
pd.read_html(url,match="Country",header=0)[0] # match -> tables are extracted based on column names inside match
# Wikipedia often blocks direct automated requests → leading to 403 Forbidden.
#Wikipedia may return HTTP 403 for automated requests; using requests with a browser User-Agent header allows pandas to read HTML tables successfully.
```
