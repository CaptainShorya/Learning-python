```python
## Assignment 1
import sqlite3
##Create a sqlite database named test.db
def create_db():
    connection = sqlite3.connect('test.db')
    connection.close() ##It safely ends the connection between your Python program and the SQLite database file.
    print("Database created successfully")

create_db()
```

```python
def create_table():
    connection = sqlite3.connect('test.db')
    cursor = connection.cursor()
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS employess(
            id INTEGER PRIMARY KEY,
            name TEXT NOT NULL,
            age INTEGER,
            department TEXT                    
        )
    ''')
    connection.commit()
    connection.close()
    print("Table create successfully")

create_table()
```

```python
##Assignment 2
import sqlite3
def insert_table(name,age,department):
    connection = sqlite3.connect('test.db')
    cursor = connection.cursor()
    cursor.execute('''
        INSERT INTO employess(name,age,department)
                   Values(?,?,?)
    ''',(name,age,department))
    connection.commit() # save changes
    connection.close() 
    print("Inserted Successfully")

insert_table("ADI",22,"ECE")
```

```python
##Assignment 3
import sqlite3
def fetch_all():
    connection = sqlite3.connect("test.db")
    cursor = connection.cursor()
    cursor.execute("Select * from employess")
    rows = cursor.fetchall()
    connection.close()
    for row in rows:
        print(row)
fetch_all()
```

```python
#In SQLite, parameterized queries use ? as placeholders and values are passed as a tuple.
import sqlite3
def fetch_all_bydepartment(department):
    connection = sqlite3.connect("test.db")
    cursor = connection.cursor()
    cursor.execute("Select * from employess where department =?",(department,)) #(department,) -> tuple , (department) -> not a tuple
    ## fetchall() works only with SELECT queries.
    rows = cursor.fetchall()
    connection.close()
    for row in rows:
        print(row)
fetch_all_bydepartment("CSE")
```

```python
##Assignment 4
import sqlite3
def update_byid(department,id):
    connection = sqlite3.connect("test.db")
    cursor = connection.cursor()
    cursor.execute("UPDATE employess SET department = ? WHERE id = ?",(department,id)) #(department,) -> tuple , (department) -> not a tuple
    connection.commit()
    connection.close()
update_byid("IT",2)

#Multiple records can be updated using WHERE IN clause and displayed using a SELECT query.
# WHERE IN -> match any value from the given list , eg-> WHERE id IN (1, 2) -> id = 1 OR id = 2(Only these values — nothing else.)
```

```python
##Assignment 5
import sqlite3
def delete_employees(id):
    connection = sqlite3.connect("test.db")
    cursor = connection.cursor()
    cursor.execute("DELETE FROM employess where id = ?",(id,))
    connection.commit()
    print("data deleted successfully")
    cursor.execute("select * FROM employess")
    rows = cursor.fetchall()
    connection.close()
    for row in rows:
        print(row)
    print("remainging data printed successfully")
delete_employees(4)
```
