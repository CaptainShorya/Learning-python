```python
##Assignment 7
import sqlite3
def insert_multiple(data):
    connection = sqlite3.connect("test.db")
    cursor = connection.cursor()
    try:
        cursor.executemany('''
            insert into employess(name,age,department)
                    values(?,?,?)
        ''',data)
        connection.commit()
        print("All data inserted successfully")
    except Exception as e:
        connection.rollback()
        print("Data not inserted properly")
        print(e)
    finally:
        connection.close()
data = [
    ('Jay', 22, 'IT'),
    ('Aryan', 19, 'IT'),
    ('Sachin', 20, 'BCA')
]
insert_multiple(data)
```

```python
import sqlite3
def insert_multiple(data):
    connection = sqlite3.connect("test.db")
    cursor = connection.cursor()
    try:
        cursor.executemany('''
            update employess
            set age = ?
            where age = ?
        ''',data)
        connection.commit()
        print("Age modified successfully")
    except Exception as e:
        connection.rollback()
        print("Error occured, transaction rollback")
        print(e)
    finally:
        connection.close()
data = [
    (21,22),
    (20,19)
]
insert_multiple(data)
```

```python
####Assignment 8
def create_table_department():
    connection = sqlite3.connect('test.db')
    cursor = connection.cursor()
    cursor.execute('''
        create table if not exists department(
        id integer PRIMARY KEY,
        name text
        )
    ''')
    connection.commit()
    connection.close()
    print("Table 'department' created successfully")
create_table_department()
```

```python
#A foreign key is a column in one table that refers to the primary key of another table to maintain referential integrity.
#Syntax : FOREIGN KEY (department_id) REFERENCES departments(id)
##This enforces rules:You cannot assign invalid department,You cannot insert department_id = 10 if it doesn’t exist

#You cannot directly modify an existing table to add a foreign key.SQLite does NOT support: ALTER TABLE ADD CONSTRAINT
#PRAGMA foreign_keys = ON is used to enable enforcement of foreign key constraints in SQLite.
## STEPS : Create new table with FK, Copy old data, Drop old table, Rename new table
import sqlite3
def add_department_foreign_key():
    conn = sqlite3.connect("test.db")
    cursor = conn.cursor()
    cursor.execute("PRAGMA foreign_keys = ON")
    cursor.execute("DROP TABLE IF EXISTS old_employess")
    cursor.execute("ALTER TABLE employess RENAME TO old_employess")
    cursor.execute("""
        CREATE TABLE employess (
            id INTEGER PRIMARY KEY,
            name TEXT NOT NULL,
            age INTEGER,
            department TEXT,
            department_id INTEGER,
            FOREIGN KEY(department_id) REFERENCES department(id)
        )
    """)
    cursor.execute("""
        INSERT INTO employess (name, age, department)
        SELECT name, age, department FROM old_employess
    """)
    cursor.execute("DROP TABLE old_employess")
    conn.commit()
    conn.close()
    print("Table 'employees' modified successfully.")

add_department_foreign_key()
#SQLite does NOT allow multiple SQL statements in one execute(). Inside one execute(ALTER... CREATE... INSERT... DROP...)❌❌
```

```python
def insert_department_and_employee(department_id, department_name, employee_id, name, age, department):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    try:
        cursor.execute('''
            INSERT INTO department(id, name)
            VALUES (?, ?)
        ''', (department_id, department_name))
        cursor.execute('''
            INSERT INTO employess (id, name, age, department, department_id)
            VALUES (?, ?, ?, ?, ?)
        ''', (employee_id,name, age, department, department_id)) 
        conn.commit()
        print("Department and employee inserted successfully.")
    except Exception as e:
        conn.rollback()
        print("Error occurred, transaction rolled back.")
        print(e)
    finally:
        conn.close()

# Test the function
insert_department_and_employee(1, 'CSE',1,'Shorya', 22, 'CSE')
```

```python
##Assignment 9
# An index is a data structure used by the database to speed up data retrieval.
# An index creates a separate lookup structure that stores: column value  →  row location, so the database does NOT scan the entire table.

## Without index: SQLite checks every row,Time complexity ≈ O(n)
## With index: SQLite jumps directly to “Rahul”,Time complexity ≈ O(log n)

def create_index_on_name():
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    cursor.execute('CREATE INDEX idx_name ON employess(name)') ##SYNTAX
    conn.commit()
    conn.close()
    print("Index on 'name' column created successfully.")
create_index_on_name()
```
