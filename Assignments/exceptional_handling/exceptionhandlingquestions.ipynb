```python
##Assignment 1 
def division(num1,num2):
    try:
        num = num1/num2
        print(num)
    except ZeroDivisionError as ex:
        print("Zero Division Error")
        print("Result = None")
    finally:
        print("execution completed")

division(10,0)
```

```python
##Assignment 2
def file_read(file_name):
    try:
        file = open(file_name,'r')
        content = file.read()
        print(content)
    except FileNotFoundError as ex:
        print("No such file exists")
    finally:
        try:
            file.close()
        except NameError:
            pass
file_read("example.txt")
```

```python
##Assignment 3
def sum_number(lst):
    try:
        total = 0
        for i in lst:
            total += i
    except TypeError as ex: # Wrong data type is used.Python understands what you want to do,but the type of data is incorrect
        print("non integer value encountered")
        total = None
    finally:
        print("Program Exceuted successfully")
    return total

print(sum_number([1,2,3,4,5]))
print(sum_number([1,2,3,4,"hello"]))
```

```python
##Assignment 4
def input_number(num):
    try:
        total = int(num)
    except ValueError as ex:  #Correct data type, but wrong value.Python understands the type,but the value itself doesn’t make sense.
        print("Non integral value")
        total  = None
    finally:
        print("Program Exceuted successfully")
    return total

print(input_number(10))
print(input_number("hello"))
```

```python
##Assignment 5
def dict_access(d,k):
    try:
        output = d[k]
    except KeyError as ex:
        print("Key does not exist")
        output = None
    finally:
        print("program excecuted successfully")
    return output
print(dict_access({1:1,2:4},2))
```

```python
##Assignment 6
def nested_exception(str):
    try:
        num = int(str) ## Value error occur here due to wrong value entered.
        try:
            num2 = 10/num
            print(num2)
        except ZeroDivisionError as ex:
            print(ex)
            print("zero division error occured")
    except ValueError as error:
        print(error)
        print("value error occured")   
    finally:
        print("program excecuted successfully") 

nested_exception('Shorya')
print()
nested_exception(10)
```

```python
##Assignment 7
def lst_access(lst,k):
    try:
        output = lst[k]
    except IndexError as ex:
        print("Key does not exist")
        output = None
    finally:
        print("program excecuted successfully")
    return output
print(lst_access([1,2,3,4,5],5))
```

```python
##Assignment 8
import requests #requests is a third-party library used for: HTTP request, API , web services, backend communication
def read_url(url):
    try:
        response = requests.get(url) #
        response.raise_for_status() #This checks HTTP status codes.If 200 -> Ok, 404 -> Not Found, 500 -> server error #It automatically raises an exception.
        return response.text ##Returns:HTML, JSON stringAPI, response text
    except requests.RequestException as e:#catches all network-related errors,such as: connection error, timeout, invalid URL, HTTP error, DNS failure
        print(f"Network Error : {e}")
        return None
    finally:
        print("Program Exceuted successfully")
#print(read_url('https://jsonplaceholder.typicode.com/posts/1'))
print(read_url('https://randomtocheck.url'))
```

```python
##Assignment 9
import json 
##Parsing → converting text into usable data
#Parse a json string means Converting a JSON-formatted string into a Python data structure (usually a dictionary or list).
def parse_json(json_string):
    try:
        data = json.loads(json_string) #Parse JSON string
        return data
    except json.JSONDecodeError as e:
        print(e)
        return None
    finally:
        print("execution completed")
print(parse_json('{"name": "John", "age": 30}')) #json data as text
```

```python
##Assignment 10
#Only classes that inherit from Exception are treated as errors by Python.
class NegativeNumberError(Exception): #if you don’t inherit from Exception ->Python does NOT recognize it as an error.
    pass

def raise_error(lst):
    try:
        for i in lst:
            if i<0:
                raise NegativeNumberError
    except NegativeNumberError as e:
        print("Negative Number Encountered")
    finally:
        print("Excetion completed")

raise_error([1,2,3,4,-5])
    
```

```python
##Assignment 12
class Calculator:
    def __init__(self,num1,num2):
        self.num1 = num1
        self.num2 = num2
    def division(self):
        try: 
            cal = self.num1/self.num2
        except ZeroDivisionError as e:
            print("Zero in denominator")
            cal = None
        finally:
            print("Excetion completed")
        return cal
obj1 = Calculator(10,0)
print(obj1.division())
```

```python
##Assignment 13
def convert(lst):
    try:
        for i in lst:
            print(int(i))
    except ValueError as e:
        print(e)
    finally:
        print("Excetion completed")
    
convert(["1","2","3","Hello"])
```

```python
##Assignment 14
def convert_to_integers(lst):
    try:
        integers = [int(i) for i in lst]
        print("Yes we done it")
    except ValueError as e:
        print("Error")
        print(f"Error: {e}")
        integers = None
    finally:
        print("Execution complete.")
    return integers
convert_to_integers(["1","2","3"])
```

```python
##Assignment 15
def write_strings_to_file(strings, filename):
    try:
        file = open(filename, 'w')
        for string in strings:
            file.write(string + '\n')
    except IOError as e:
        print(f"Error: {e}")
    finally:
        try:
            file.close()
        except NameError:
            pass
write_strings_to_file(['Hello', 'World'], 'output.txt')
```
