```python
##Assignment 1 teaches me -> How does python iterate itself ?
for i in range(5):
    print(i)
## Python uses an iterator, but python hides the mechanism behind it (shown below)
iterator = iter(range(5)) ## 0 | 1 | 2 | 3 | 4 , In starting iterator points over 0 and after providing value to i, pointer moves forward
while True:
    try:
        i = next(iterator) ## return one value at a time
        print(i)
    except StopIteration:
        break

```

```python
##Assignment 1
class Counter:
    def __init__(self,count): ## store starting value
        self.current = count
    
    def __iter__(self): ## tell Python this is iterable
        return self
    
    def __next__(self): ## return next value
        if self.current < 0:
            raise StopIteration() ## stop loop
        else:
            value = self.current
            self.current -= 1
            return value
    
n = 10
for i in Counter(n):
    print(i)
```

```python
##Assignment 2
class MyRange:
    def __init__(self,num):
        self.i = 0
        self.current = num

    def __iter__(self):
        return self
    def __next__(self):
        if self.i < self.current:
            value = self.i
            self.i += 1
            return value
        else:
            raise StopIteration
        
for i in MyRange(5):
    print(i)
```

```python
##Assignment 3
## Generator automatically raises: StopIteration
def fibonacci(num):
    n1= 0
    n2 =1
    for _ in range(num):
        yield n1 ##In iteration 2 function begins right after yield
        n1,n2 = n2, n1+n2

for num in fibonacci(10):
    print(num)
```

```python
##Assignment 5
##Even number
def even_numbers(num):
    for i in range(1,num+1):
        if i%2 ==0 :
            yield i
##Square number
def squares(numbers):
    for number in numbers:
        yield number * number
##formation of generators
even_gen = even_numbers(20)
sq_gen = squares(even_gen)
for eve_sq in sq_gen:
    print(eve_sq)

##Using generator expression 
'''nums = (i for i in range(10))
sq = (i*i for i in nums)
ev = (i for i in sq if i % 2 == 0)

for x in ev:
    print(x)'''
```

```python
##Assignment 6
import time

def time_it(func):
    def wrapper(*args,**kwargs):
        start = time.time()
        result = func(*args,**kwargs)
        end = time.time()
        print(f"Execution time :{end-start} seconds")
        return result
    return wrapper

@time_it
def factorial(n):
    fact = 1
    for i in range(1,n+1):
        fact *= i
    return fact

print(factorial(6))
```

```python
##Assignment 7
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator

@repeat(3)
def print_message(message):


    
    print(message)

print_message("Hello, World!")
```
